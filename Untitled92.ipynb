````python
import os
os.environ['PYTORCH_CUDA_ALLOC_CONF'] = 'expandable_segments:True'
# Translation task
tgt_lang = "Telugu"
input_txt = r"""# Spooky Memories in Higher Dimension (Superposition)

This section is a replication of [Toy Models of Superposition](https://transformer-circuits.pub/2022/toy_model/index.html). Clean interpretations can give us a sense of security. If you've worked on deploying models in mission-critical scenarios, you can't help but be frustrated when things don't fall into a human-understandable realm. The ideal scenario would be for individual neuron activations to map cleanly to individual features.

Several works have attempted to quantify this phenomenon. We have observed that there are several neurons that map cleanly to features, but the reverse — features mapping cleanly to neurons — is much rarer and harder to find.

There is a particularly useful analogy: neural networks represent features in activation space as directions. When we state things this way, we are making a much stronger claim about their representations. There is significant empirical evidence and argument supporting this claim, but for it to hold, the properties of decomposability and linearity must be true.

If features correspond directly to neuron activations, identifying them becomes a lot easier. If not — i.e., if they're in superposition — they become a lot harder to grasp. This push and pull can be decomposed into two distinct forces: **privileged basis** (clean neuron activations where features align to the basis) and **superposition** (the network trying to simulate a larger network).

Embedding arithmetic has been observed to hold true, further cementing the above analogy. If we consider gender as a direction in embedding space, then v(KING) − v(man) + v(woman) = v(QUEEN). Interpretable directions have also been found across several different architectures. Phenomena like universality and polysemanticity have also been observed in language models — much of what we've observed in vision models transfers here.

How do we even define a feature? If they represent tokens, then every new operation should open up a new representation. If we take the arithmetic example above, "king − man" can be thought of as an operation on individual features — this is the functional representation. We can also think of features as human-understandable properties, though novel concepts can't always be represented this way. Another analogy for representing features is to think of them as a list of rules, like "activation in neuron 1 is connected to neuron 7 in layer 2," and so on.

![[Pasted image 20260723172617.png]]

| rule number | rule code | explanation                                                                     |
| ----------- | --------- | ------------------------------------------------------------------------------- |
| rule 1      | h1        | the neuron activation value for the l11 will be set to (intensity of the color) |
| rule 2      | h2        | the neuron activation value for the l12 will be set to (intensity of the color) |
| rule 3      | h3        | the neuron activation value for the l13 will be set to (intensity of the color) |
| rule 4      | h4        | the neuron activation value for the l14 will be set to (intensity of the color) |

The hope with this approach is that, given a sufficiently large neural network, it will dedicate a neuron to a single concept — e.g., neuron l11 representing "being a man," etc.

In a linear representation, the presence of multiple features is represented as $x_{f_{1}}W_{f_{1}} + x_{f_{2}}W_{f_{2}} + x_{f_{3}}W_{f_{3}} \dots$, where $W$ represents the representation direction and $x_{f}$ represents the activation of $x$ along that particular direction.

![[Pasted image 20260725153628.png]]

Here, $x_{f_i}$ can take any magnitude — e.g., how happy someone was, or whether they won first place, etc.

So why should the model align with these specific directions to represent concepts? If there's an incentive for features to align with basis dimensions, we call it a **privileged basis**; if not, we call it a **non-privileged basis**. (In some sense, after training, concepts like king, man, boy, father, uncle, prince, etc. emerge this way — it's much easier if gender is a direction.)

Even with a privileged basis, neurons are sometimes polysemantic — activating for seemingly unrelated features. We call this superposition.

> **Superposition** — when a model represents more features than it has dimensions.

When trying to represent more features than it has neurons, polysemanticity emerges. This is mathematically plausible — one way is by using almost-orthogonal vectors (there are exp(n) almost-orthogonal vectors in n dimensions), and another is compressed sensing (recovering sparse signals from lower-dimensional projections). These representations can interfere with each other, but since sparsity is so high in larger and larger dimensions, the cost may outweigh the benefit.

![[Pasted image 20260725165107.png]]

![[Pasted image 20260725165346.png]]

The yellower they are, the higher the importance of the feature, and the most important features are projected onto orthogonal and antipodal pairs.

We pick a total of 5 features, with importance and feature probability calculated as:

$$
\text{importance}_{i} = (0.9)^{i}, \qquad i = 0, 1, \dots, n_{\text{features}}-1
$$
$$
\text{feature\_probability}_{j} = 20^{-t_j}, \qquad
t_j = \frac{j}{n_{\text{instances}}-1}, \quad j = 0, 1, \dots, n_{\text{instances}}-1
$$

and project them onto 2 dimensions with increasing levels of sparsity.

A simple but powerful way to understand what's going on here is to pick a much smaller, less complex model and study it directly. We saw in the induction heads section that more layers does indeed lead to generalization.

To visualize this, we use a small ReLU network trained on synthetic data with sparse input features to demonstrate superposition.

#### ReLU Network

The sparse vectors are generated from a Bernoulli distribution.

![[Pasted image 20260725064818.png]]

The model is a ReLU encoder with a bottleneck of d < D:
$$
h=ReLU(Wx)=ReLU\left( \sum_{i-1}^D \u200b x_{i}w_{i}\u200b \right)\in R^d.
$$
Here, $w_1, w_2, \dots, w_D \in \mathbb{R}^d$ are the embedding vectors for each feature.

![[Pasted image 20260725071532.png]]

When features are sparse, superposition allows compression beyond what a linear model would do, at the cost of interference that requires nonlinear filtering. In the following sparse vector generation process, we use a slight variation on the Bernoulli distribution: instead of picking values from just $\{ 0,1 \}$, we pick values from an inflated uniform distribution. Here, the presence of a feature is determined by a Bernoulli trial, with the probability being the feature probability.

$$
x_i =
\begin{cases}
U_i \sim \mathcal{U}(0,1), & \text{with probability } p_i^{(k)} \\
0, & \text{with probability } 1-p_i^{(k)}
\end{cases}
$$

We observe that superposition emerges only in models with a nonlinearity (ReLU); for linear models, there is simply no interference.

$$
\begin{aligned}
&\textbf{Linear Model} \\
&\mathbf{h} = \mathbf{x} \mathbf{W} \qquad (\mathbf{x}\in\mathbb{R}^{1\times d},\; \mathbf{W}\in\mathbb{R}^{d\times h},\; \mathbf{h}\in\mathbb{R}^{1\times h}) \\
&\hat{\mathbf{x}} = \mathbf{h} \mathbf{W}^{\!} \top + \mathbf{b} = \mathbf{x} \mathbf{W} \mathbf{W}^{\!} \top + \mathbf{b}
\end{aligned}
$$

For linear models, there is almost no superposition and no interference, but as soon as nonlinearity is introduced, this changes. We took a total of 100 different features, with 20 hidden dimensions and 20 instances, for the plot below — the yellow portion represents superposition. We represented $W^TW$, with the blue dots in the right-side plots representing superposition; as we can see, superposition increases with increasing sparsity.

$$
\begin{aligned}
&\textbf{ReLU Output Model} \\
&\mathbf{h} = \mathbf{x} \mathbf{W} \\
&\hat{\mathbf{x}} = \operatorname{ReLU}\!\big(\mathbf{h} \mathbf{W}^{\!} \top + \mathbf{b}\big) = \operatorname{ReLU}\!\big(\mathbf{x} \mathbf{W} \mathbf{W}^{\!} \top + \mathbf{b}\big)
\end{aligned}
$$

Below is the resulting plot:

![[newplot.png]]

The linear model always learns the top-m most important features, analogous to learning the top principal components. The ReLU output model behaves the same way on dense features, but as sparsity increases, we see superposition emerge.

$$
\begin{aligned}
&\textbf{Loss (both models)} \\
&\mathcal{L} = \sum_{\mathbf{x}} \sum_{i=1}^{d} I_i\,(x_i - \hat{x}_i)^2, \quad I_i \ge 0 \text{ (feature importance)}
\end{aligned}
$$

The learning dynamics amount to a tug-of-war between two phenomena: feature benefit and interference. There's a decrease in loss from representing a feature, offset by potential interference from non-orthogonal features affecting each other's predictions.

$$
% Equation 1:
L_1 = \sum_i \int_{0 \le x_i \le 1} I_i \left(x_i - \text{ReLU}(\|W_i\|^2 x_i + b_i)\right)^2 + \sum_{i \neq j} \int_{0 \le x_i \le 1} I_j \text{ReLU}(W_j \cdot W_i x_i + b_j)^2
$$

$$
% Equation 2:
L_1 = \sum_i I_i \left(1 - \text{ReLU}(\|W_i\|^2 + b_i)\right)^2 + \sum_{i \neq j} I_j \text{ReLU}(W_j \cdot W_i + b_j)^2
$$

These equations primarily consist of two terms, as explained earlier: the feature benefit term, which penalizes the model if it fails to reconstruct a target feature, and the second term, which penalizes the model if activating feature $i$ accidentally activates another feature $j$ — it measures the overlap.

The following three things can happen to a feature:

1. The feature is learned with a dedicated dimension
2. The feature is learned in superposition
3. The feature is simply not learned

### Phase Diagram

One of the better ways to understand which category a feature falls into is by using a phase diagram. As established earlier, based on importance and sparsity levels, features can fall into any of the presented states — we call these states dropped, superposed, and dedicated. In the phase diagram, one of the most striking things we've noticed is how abrupt the change is when training conditions are varied: the boundary between the three states looks sharp rather than gradual, much like how water doesn't gradually "soften" into steam but flips abruptly at 100°C. When observing this phenomenon in large models, the effects seem much more tangled. To strip away the complexity, we looked at four smaller models mapping from 3D to 2D, picking two features, $x_1$ and $x_2$, and generating them using the uniform inflated distribution presented above. We also assigned a relative importance $r$ to one feature, since some features are much more important than others — we picked values anywhere from 0.1 to 10 for $r$, while fixing the value for the other feature. We also varied $s$ (sparsity) across the range 1 to 0.01. The loss looks like this:

$$
\tloss = (x\u2081-y\u2081)² + (x\u2082-y\u2082)² + ... + (x_{n-1}-y_{n-1})² + r·(x_n-y_n)²
$$

where **r** signifies that some features are more important than others.

The four models we picked are simple, each of the form $y = \text{ReLU}(W^TWx + b)$, where $\ker(W)$ has dimension 1. To map into lower dimensions, we either ignored coordinates or superimposed them.

For generating the above 3 cases we used the following procedure:
- **Discard-and-guess.** Zero out a coordinate, but add back its *unconditional mean*
  (`s/2`, since it's `0` with probability `1-s` and averages `0.5` when active) as a
  constant bias. This is the best *constant* guess for a coordinate you've thrown away.
- **Superposition.** Apply the 2\u00d72 matrix
  $$
  P_2=\begin{pmatrix}
1 & -1\\
-1 & 1
\end{pmatrix}
$$
  to a *pair* of coordinates, then ReLU. Geometrically this is a projection orthogonal to
  `(1,1)`. If only one of the two coordinates is active, it comes back out perfectly
  (e.g. `(x, 0) \u2192 (x, -x) \u2192 ReLU \u2192 (x, 0)`). If **both** are active at once, they
  interfere — only the *difference* survives — and that mismatch is the cost of
  superposition.

The four models:

| Name        | What it does                                                    | Bottleneck direction thrown away    |
| ----------- | --------------------------------------------------------------- | ------------------------------------ |
| **f**       | discard the **first** coordinate (weight 1), add back its mean  | 1st coordinate                      |
| **g**       | discard the **last** coordinate (weight `r`), add back its mean | last coordinate                     |
| **h_first** | **superpose** the first two coordinates                         | interference between coords 1 & 2   |
| **h_last**  | **superpose** the last two coordinates (one has weight `r`)      | interference between coords n-1 & n |

`f` and `g` are mirror images of each other, as are `h_first` and `h_last` — the only
difference is *which* coordinate carries the special weight `r`.

For a given sparsity parameter $s$, we will consider four tiny "neural networks." The first two include small bias terms that reflect the expected value of the coordinate being discarded. (Here $e_1$ and $e_n$ are the first and last basis vectors.)

$f_s(x) = \text{ReLU}(D_{first}x  + \frac{\displaystyle s}{\displaystyle 2}e_1)$

$g_s(x) = \text{ReLU}(D_{last}x  + \frac{\displaystyle s}{\displaystyle 2}e_n)$

$h_{first}(x) = \text{ReLU}(S_{first}x)$

$h_{last}(x) = \text{ReLU}(S_{last}x)$

To better understand the relationship between loss and activity, we manually picked particular $x$ values and studied various cases to see where each point falls, fixing $(s = 0.3, r = 2.0)$, to be displayed in the phase diagram.

### Case A — only the (unweighted) middle coordinate is active: `x = (0, 0.55, 0)`

| model                   | output y          | loss       |
| ----------------------- | ------------------ | ---------- |
| f (discard first)       | `(0.15, 0.55, 0)` | 0.0225     |
| g (discard last)        | `(0, 0.55, 0.15)` | 0.0450     |
| h_first (superpose 1,2) | `(0, 0.55, 0)`    | **0.0000** |
| h_last (superpose 2,3)  | `(0, 0.55, 0)`    | **0.0000** |

Nothing collides here, so both superposition models reconstruct perfectly, while `f`
and `g` still pay a small penalty for guessing at a coordinate that happens to be zero
(they don't know that in advance).

### Case B — only the first coordinate is active: `x = (0.72, 0.10, 0)`

| model                   | output y             | loss       |
| ----------------------- | --------------------- | ---------- |
| f (discard first)       | `(0.15, 0.10, 0)`    | 0.3249     |
| g (discard last)        | `(0.72, 0.10, 0.15)` | 0.0450     |
| h_first (superpose 1,2) | `(0.62, 0, 0)`       | 0.0200     |
| h_last (superpose 2,3)  | `(0.72, 0.10, 0)`    | **0.0000** |

`f` pays heavily here — it threw away exactly the coordinate that was active.
`h_first` collides `x\u2081` with the (small) `x\u2082`, but the damage is small since `x\u2082` is
small. `h_last` is untouched since it only entangles coordinates 2 and 3.

### Case C — only the last (weighted) coordinate is active: `x = (0, 0.10, 0.85)`

| model | output y | loss |
|---|---|---|
| f (discard first) | `(0.15, 0.10, 0.85)` | 0.0225 |
| g (discard last) | `(0.10, 0.10, 0.15)` | **0.9800** |
| h_first (superpose 1,2) | `(0, 0.10, 0.85)` | **0.0000** |
| h_last (superpose 2,3) | `(0, 0, 0.75)` | 0.0300 |

`g` is disastrous — it discarded the coordinate that matters twice as much, *and* it was
active. This is the key asymmetry that makes `r` matter.

### Case D — both special coordinates active (a "collision"): `x = (0.72, 0.10, 0.85)`

| model                   | output y             | loss   |
| ----------------------- | --------------------- | ------ |
| f (discard first)       | `(0.15, 0.10, 0.85)` | 0.3249 |
| g (discard last)        | `(0.72, 0.10, 0.15)` | 0.9800 |
| h_first (superpose 1,2) | `(0.62, 0, 0.85)`    | 0.0200 |
| h_last (superpose 2,3)  | `(0.72, 0, 0.75)`    | 0.0300 |

Here `h_first` is cheapest: it's better to let the two *unweighted* coordinates
interfere a little than to mishandle the weighted one at all.

The plot below shows the phase diagram; the x represents $(s = 0.3, r = 2.0)$:

![[newplot (1).png]]

The log loss surface, capturing loss as a function of sparsity and relative weight, is shown below:

![[newplot (3).png]]

The diagram below shows the log loss of the four selected models:

![[newplot (4).png]]

## Uniform and Nonuniform Superposition

Uniform superposition is where all the features are equally sparse and equally independent, and there is no interference.

Since we are considering equally important features (i.e., I = 1) and equal sparsity, an interesting thing to look at is not the number of features $n$ or the number of hidden features $m$, but the ratio between them.

So how do we even measure how many features are actually learned? To do this, we use the Frobenius norm $||W_i||^2$: for any represented feature this would be 1, and for any non-represented feature it would be 0. This measure is basis-independent and remains valid even in the dense regime.

The weight matrix of the model has shape (m, n), as defined above. If n >> m, the model cannot give every feature its own orthogonal direction in the lower-dimensional space — it must superpose multiple features.

So we need a measure of how many dimensions are devoted to each feature, since, as we've seen, orthogonality isn't always possible. Using the metric defined earlier, $||W||_{F}^2$ = 1 when each dedicated feature gets its own orthogonal dimension. From this, we arrive at a summary statistic called $D^*$, the average number of dimensions devoted to a single learned feature:

$$
D^* = m / ||W||_F^2
$$

**`D* = 1` — every feature gets its own dedicated dimension** (no superposition)

With `n = 3` features and `m = 3` hidden dimensions, each feature mapped to its own orthogonal direction:

```
                hidden1  hidden2  hidden3
   feature 1  [    1        0        0   ]
   feature 2  [    0        1        0   ]
   feature 3  [    0        0        1   ]
```
$$
||W_{dedicated}||_{F}^2 = 1² \times 3 = 3
$$$$
D^* = m / ||W||_F² = 3 / 3 = 1$$

**Case 2 — D\* = 1/2: antipodal pairs**

With `n = 6` features sharing only `m = 3` hidden dimensions, two features per dimension pointing in opposite directions (`+1` / `-1`):

```
                hidden1  hidden2  hidden3
   feature 1  [    1        0        0   ]
   feature 2  [   -1        0        0   ]
   feature 3  [    0        1        0   ]
   feature 4  [    0       -1        0   ]
   feature 5  [    0        0        1   ]
   feature 6  [    0        0       -1   ]
```

$$||W_{antipodal}||_{F}^2 = 1² \times 6 = 6$$$$
D^* = m / ||W||_{F}^2 = 3 / 6 = 1/2$$

This phenomenon is similar to the fractional quantum Hall effect. I've plotted the two curves so you can grasp the parallels. The quantum Hall effect occurs when a 2D electron gas, at low temperature and high magnetic field, shows Hall conductance quantized into exact integer or fractional multiples of e²/h. These plateaus arise from topological properties of the electron wavefunctions, which form Landau levels; the fractional version stems from strong electron correlations creating quasiparticles with fractional charge. It's a fascinating phenomenon, and looking into it as a slight detour would make for a great afternoon read.

![[newplot (5).png]]

In the fractional $D^* = 1/2$ case above, the features come in antipodal pairs, $W_k = -W_j$ — exact negatives.

These antipodal pairs raise an interesting question: how can per-feature dimensionality be distributed more finely? To answer this, we define the dimensionality of the $i$th feature, $D_i$, as:

$$
D_i = \frac{\|W_i\|^2}{\sum_j (\hat{W}_i \cdot W_j)^2}
$$

The numerator represents to what extent the feature is represented, and the denominator represents how many features share the dimension it's embedded in.

In the antipodal pairs case, we get a nice $D_i = 1/2$. The graph is "sticky" — in this context, stickiness refers to the following: if a feature is sticky, the network keeps using it even after transformations or layer changes; if a feature is not sticky, its representation changes more freely and the model becomes less dependent on any single dimension.

![[newplot (7).png]]

If we take these generated superposition models and vary the sparsity levels, we observe some features go from getting absolutely no representation to getting a fully dedicated dimension. Watching how these phase changes occur produces quite an interesting plot, similar to electron phase diagrams akin to Landau levels — we see interesting geometric patterns emerge, structures like digons, triangles, and tetrahedra, which minimize energy in a way reminiscent of chemistry.

<iframe src="https://factity.github.io/interpretability/plotsiframe/phase.html" width="100%" height="600" style="border:0;" loading="lazy"></iframe>

This tells us that the fractional dimensionality we've observed is indeed superposition. When our model chooses to represent a feature, it represents it on an m-dimensional sphere; since we're using uniform importance, these representations tend toward symmetry.

The core mathematical phenomenon at play here is the equivalence between point configurations and Gram matrices. Given n features represented in m dimensions:

Any set of $n$ points in $\mathbb{R}^m$ gives a rank-$m$, $n \times n$ positive semi-definite matrix whose entries are the dot products between the points.

The matrix $W^TW$ captures how much these features interfere with each other. The diagonal entries of this matrix represent the individual features, and the off-diagonal entries represent the interference between features in representational space.

The diagram below shows how features are arranged in 2D space:
<iframe src="https://factity.github.io/interpretability/plotsiframe/2d.html" width="100%" height="600" style="border:0;" loading="lazy"></iframe>
The diagram below shows how features are arranged in 3D space:
<iframe src="https://factity.github.io/interpretability/plotsiframe/3d.html" width="100%" height="600" style="border:0;" loading="lazy"></iframe>

## Nonuniform Superposition

In most real-world networks, uniform superposition is rarely observed.

With non-uniform superposition, there's a smooth deformation of the polytopes we discussed earlier — akin to molecules under increasing energy, as more and more imbalance builds up, the polytopes break into smaller ones. Correlated features were also observed to fall into orthogonal directions, essentially forming a local basis, while anti-correlated features interfere with each other and fall in almost the same direction.

![[newplot (6).png]]

If two features, a and b, always fire together (or are both zero), and you give the model only one dimension to represent them, the model maps a and b into a single number z.

- **PCA solution** – finds the direction of maximum variance.
    For two perfectly co\u2011activating features, the first principal component is $\frac{a+b}{\sqrt{2}}$, and the second, which captures their difference, is $\frac{a-b}{\sqrt{2}}$.
    With only one dimension, PCA keeps the **sum** and throws away the **difference**. So the model represents the shared component and ignores what makes them distinct. This is the **collapse**: two features merge into their principal component.

- **Superposition solution** – exploits _sparsity_ (features are rarely active).
    If a and b are active only very rarely, the model can represent both in the same dimension by assigning them **almost orthogonal directions** in the high\u2011dimensional neuron space — here, just opposite signs (e.g., a \u2192 +1, b \u2192 -1). Because they rarely fire together, their interference is minimal, so the model can recover each one individually despite the bottleneck.

As the level of sparsity decreases, superposition gradually decreases too, and the representation becomes equivalent to PCA.

If we plot similar toy models to the ones above, but now with correlation included, and gradually increase the sparsity, the following structural collapse happens. I've made visualizations of this phenomenon in 2D and 3D — focus specifically on the dotted lines in the plots below.

<iframe src="https://factity.github.io/interpretability/plotsiframe/2dcollapse.html" width="100%" height="600" style="border:0;" loading="lazy"></iframe>

<iframe src="https://factity.github.io/interpretability/plotsiframe/3dcollapse.html" width="100%" height="600" style="border:0;" loading="lazy"></iframe>

<iframe src="https://factity.github.io/interpretability/plotsiframe/featurecollapse.html" width="100%" height="600" style="border:0;" loading="lazy"></iframe>

You can control this manually using the slider below; there are two different variables changing here:

<iframe src="https://factity.github.io/interpretability/plotsiframe/tegumcollapse.html" width="100%" height="600" style="border:0;" loading="lazy"></iframe>

Unlike what one might expect, a fully trained neural network converges to a simple but nontrivial structure that rhymes with geometry, rather than an arbitrary sequence of weights like the simple examples presented earlier. The way this property emerges is more discrete than continuous, as seen in the phase plot.

If we plot the learning dynamics of the model against dimensionality, we observe several jumps in the loss curve. This can be explained by the phenomenon of grokking, where there's an abrupt jump. One nice analogy is the phase transition of water into gas: when you heat water, it doesn't gradually turn into gas — it gains energy until a particular temperature is reached, and then suddenly starts turning to gas. The continuity in most regions of the curve is due to a similar phenomenon.

## Learning as a Geometric Transformation

We can see in the plot below how correlated features occupy orthogonal pairs, while anti-correlated ones move in opposite directions. As training progresses, we can see the features move toward geometrically stable positions.

<iframe src="https://factity.github.io/interpretability/plotsiframe/geometrictransformation.html" width="100%" height="600" style="border:0;" loading="lazy"></iframe>

As sparsity increases, more and more features become intertwined in superposition. To help grasp how activations are affected, and how a feature contributes to an activation, I took a sample of 5 neurons and 7 features to show how they're mapped, and how the representations in superposition change over time. What starts out as fairly good separation moves toward a spectrum.

<iframe src="https://factity.github.io/interpretability/plotsiframe/neuronlevelview.html" width="100%" height="600" style="border:0;" loading="lazy"></iframe>

## Relationship to Adversarial Robustness

The real question that arises is: how can we ensure the model is robust? How do we make sure the model doesn't hide harmful thoughts and features in superposition?

<iframe src="https://factity.github.io/interpretability/plotsiframe/reconstructattack.html" width="100%" height="600" style="border:0;" loading="lazy"></iframe>

In a model without superposition, the end-to-end weights for a feature are

$(W^TW)  =  (1, 0, 0, 0, ...)$, while for a model with superposition they look like $(W^TW)  =  (1, \epsilon, -\epsilon, \epsilon, \dots)$.

This opens up a whole new avenue of attack. The network is fairly robust with normal training — it only breaks when attacks are specifically targeted at a particular feature, using an optimal L2 attack for each feature: $(\lambda(W^TW)_{i}/\u2223\u2223(W^TW)_{i}\u2223\u2223\u200b)$.

This is one way a perpetrator could hide harmful features — not the only way, but superposition could explain a lot of the reasons a model might store them this way. In other words, if we look at superposition, we can say the harmful phenomenon resides within it. To understand how the attack matters, and at which stage of training, I've plotted the target feature under attack vs. sparsity level:

<iframe src="https://factity.github.io/interpretability/plotsiframe/sparsitysweepattack.html" width="100%" height="600" style="border:0;" loading="lazy"></iframe>

We've also plotted the target under attack against training stage, overlapping the training loss on the same curve. As we can see in the plot below, attacks at the earlier training stages allow the model to hide the attack surface well and cause much more severe consequences:

<iframe src="https://factity.github.io/interpretability/plotsiframe/trainingsweep.html" width="100%" height="600" style="border:0;" loading="lazy"></iframe>

As we've observed earlier, the attack surface grows large because the network doesn't just represent features — it also performs computation in superposition. I've plotted computation between two features and how it's represented in the neurons.

/// generate the plot for this

## What This Means for AI Safety

Ultimately, we want to have a grasp on all the features being represented — we want to detect any form of deceptive tendencies happening inside the model. We want universal quantifiers over the model: we want to decompose the activation space and land on a basis we can rely on, describe activations in terms of features, and be able to explain any phenomenon we observe, while understanding and mapping out the weights. Although we attribute tremendous performance gains to polysemanticity, we also attribute much of our unfamiliarity with large models to it. We can avoid this either by removing superposition entirely, using an overcomplete basis, or some mix of the two — but once we scale models even slightly, it becomes surprisingly hard to grasp what's going on.

If we remove all forms of superposition, we end up with much cleaner models, but at a significant performance cost. One way to implement this at a higher level is to use mixture-of-experts models — we can assign blocks to features, and certain regions of the model activate only when performing certain tasks. Although this is a good alternative, at the model scales we have today we can't fully endorse getting rid of superposition altogether.

We can also look at using an overcomplete basis, but training a sparse-autoencoder-like model on top of a billion-parameter model becomes almost impossible — we'd need thousands of times the number of neurons to represent all the features, and the interference works against this too. In the section below, we'll cover monosemanticity, look at which concepts appear, and run some experiments on them.

Things we hope to take away from the following post:
- [ ] Superposition is real and an observed phenomenon
- [ ] Neurons can be both polysemantic and monosemantic
- [ ] Some form of computation can be performed in superposition
- [ ] Phase change dictates whether features are stored in superposition
- [ ] We can represent features as geometric structures

## Monosemanticity

In the previous sections, we've seen the phenomenon of superposition, along with both its positive and negative effects.

As the dimensionality of the features grows larger, sparsity grows rapidly, and distances become less informative.

![[Excalidraw/Pasted image 20260722143150.png]]

The volume of the latent space grows exponentially larger. The internal states can't be understood unless they're decomposed into independent components.
![[Excalidraw/Pasted image 20260722151016.png]]

The concept of a local neighborhood $\epsilon$ becomes less meaningful. Things are a lot more interpretable in single-layer, attention-only models, but as size grows and MLP blocks are added, it becomes almost impossible to break the model down into interpretable features.

![[Excalidraw/Pasted image 20260722164401.png]]

So the natural question arises: how do we approach breaking down the MLP block to understand what's going on inside the model?

In the previous sections, we covered simpler architectures with one or two layers, and looked at induction heads inside [transformers](https://factity.github.io/safe-ai/interp_test). We focused exclusively on attention blocks, breaking them down into one-layer and two-layer attention-only models, and looked at how bigram statistics emerged. When it comes to the MLP block, it's impossible to break it down in this same way, because the MLP is a traditional neural net where all layers matter — to put it simply, if we only took one layer's depth, the model wouldn't predict well. We need all the different layers together to get the performance needed for an accurate prediction. Fortunately, traditional neural nets already have established interpretability frameworks and a large body of work covering architectures like CNNs, RNNs, GANs, etc.

Our objective is to break down the MLP block and understand which concepts are associated with which neurons, so we map $\mathbf{x}^j$ to different activations in the layer.

$$
\mathbf{x}^j \approx \mathbf{b} + \sum_i f_i(\mathbf{x}^j) \mathbf{d}_i
$$

![[Pasted image 20260723073804.png|510]]

Here, $d_i$ is a unique term associated with a direction in feature space. This process is called linear vector factorization, and each direction is associated with a concept. To understand the activations, we train a sparse autoencoder on the activation layer:

$$
f_i(x) = \text{ReLU}( W_e (\mathbf{x} - \mathbf{b}_d) + \mathbf{b}_e )_i,
$$

where $W_e$ is the weight matrix of the encoder, and $\mathbf{b}_d, \mathbf{b}_e$ are the pre-encoder and encoder biases, respectively. The feature directions are the columns of the decoder weight matrix $W_d$.

![[Excalidraw/Pasted image 20260722181138.png]]

From the section above on superposition, we've seen that neurons try to represent more features than they have dimensions, essentially forming an overcomplete basis. To decompose the activation layer using the sparse encoder and interpret the results, we need the following to be satisfied:

1. We should have a clear understanding of which inputs lead to the required activations. This property relates to causality, generality, and purity, and helps us grasp which property activates a specific instance.

2. Downstream layers should pass on the effects of any perturbation. If we make a simple change in an upper layer, the change should be reflected in subsequent layers.

3. Features should explain a significant portion of the MLP's functionality.

If we're able to satisfy the above criteria, we can:

- Determine how much a feature changes the output and the activations of the next layer
- Monitor the network for the activations of a specific layer
- Observe a trickle-down effect, i.e., changing upper layers changes downstream layers
- Confirm the network actually works, i.e., that it's able to approximate the data
- Confirm the data is actually influencing the output
- Design inputs that will trigger a specific activation [^1]

\tEven if we were able to remove superposition completely, we'd still observe it — no amount of architectural change guarantees its complete removal. Suppose we took a single neuron with 4 mutually exclusive features and one-hot encoded them to eliminate any overlap (A/B/C/D); we'd still observe superposition, simply from the inherent nature of mutual exclusivity. The loss incentivizes the model to acquire superposition anyway, since it reduces the overall loss.

![[Pasted image 20260723090450.png]]

Superposition shows that a large set of sparse features can be represented in a lower-dimensional space. Natural latent variables are sparse, and we're trying to map a higher-dimensional vector from a lower-dimensional projection. Greedy methods can approximate this reasonably well, but sparse autoencoders work well largely because of their scalability — they can be scaled up to approximate any large dataset.

![[Pasted image 20260723105729.png]]

The sparse encoder we've designed has a linear layer feeding into the encoder block with a ReLU activation, followed by a decoder block with another linear layer.

The best factorization is the one that minimizes the total information content of the autoencoder and the data.

The importance of scale: training models on larger amounts of data has been shown to form sharper boundaries and increase expressivity.

Some neurons don't activate during training and appear dead; through resampling, we can rejuvenate these dead neurons. Once that's done, the next logical step is measuring the performance of the autoencoder. We can understand how well it works by manually checking for patterns, by analyzing feature density to find which tokens the model fires on, by looking at the MSE loss to see how accurately it maps the MLP, and so on.

### Advantages of Using a One-Layer Model

- Fewer features to represent, i.e., we're able to cover all the features using the dictionary we're using.
- We're able to train the models quite cheaply; a higher number of training tokens allows for cleaner representations.
- Because of the linearity, we can easily analyze the logit outputs, and we've shown that these properties generalize to multilayer models as well — giving us the opportunity to analyze not just the data distribution but also the functionality of the model.

///////////////////////////////////////////////////////////////////////////////////////////////////////////////////

What parameters do we need to keep in mind?

--> Find the model from which the features originated
--> The learned factors and L1 coefficients used
--> The final feature index corresponding to the final feature in the run

![[Pasted image 20260723160055.png]]

Interface for explaining the features

/////////////////////////////////////////////////////////////////////////////////////////////////////////////////////

Investigation of the individual features and some of the findings:

text written in Arabic | DNA sequences | base 64 strings | text written in Hebrew

The following will be established for the learned features:

- The learned feature activates with high specificity for the hypothesized context
- The learned feature activates with high sensitivity for the hypothesized context
- The learned feature causes appropriate downstream behavior
- The learned feature does not correspond to a neuron
- The learned feature is universal

![[Pasted image 20260723164424.png|697]]

After the experiments are done, list what you've observed about which activations caused what, exploring the following.

Select a set of characteristics and, for each one, examine:
1. Activation specificity
2. Activation sensitivity
3. Downstream effects
4. Whether the feature is not a neuron
5. Universality

//////////////////////////////////////////////////////////////////////////////////////////////////////////////////////

### Global Analysis

In the section above, we associated particular features with their activations. In this section, we'll look at the overall picture.

//////////////////////////////////////////////////////////////////////////////////////////////////////////////////////

Note the number of features the selected model, with a fixed L1 coefficient, will display, and find how many of them are dead or ultra-low-density, displaying atypical properties.

Look at a typical feature and how interpretable it is, using both human analysis and two automated interpretability methods.

### Feature Types

After modeling, we found two categories: context features and token-in-context features.

From a theoretical standpoint, the transformer takes in tokens and produces an output; n-token conjugations occur in the MLP blocks.

As a token-in-context feature example, the word **the** was activated across 100 different features, each in a different context.

|                   | Transformer | sparse autoencoder |
| ----------------- | ----------- | ------------------- |
| layers            |             |                     |
| neurons           |             |                     |
| dataset           |             |                     |
| loss              |             |                     |
| number of weights |             |                     |
| algorithm         |             |                     |

#### Key Takeaways

- Sparse autoencoders extract relatively monosemantic features.
- Sparse autoencoders produce features that aren't readily associated with the neuron basis.
- Once we've associated patterns with the autoencoder's features, we can trigger it to generate specific outputs.
- The features found with autoencoders are universal.
- The wider the autoencoder, the greater our ability to capture more concepts and finer-grained structure.
- A relatively small number of neurons can represent complex concepts.
- Generate finite state automata diagrams for what is happening."""

# Chat-style message prompt
messages = [
    {"role": "system", "content": f"Translate the text below to {tgt_lang}. and do not change the content preserve as much as you can variable names should not be translated and write it as markdown code"},
    {"role": "user", "content": input_txt}
]

# Apply chat template to structure the conversation
text = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True
)

# Tokenize and move input to model device
model_inputs = tokenizer([text], return_tensors="pt").to(model.device)

# Generate the output
generated_ids = model.generate(
    **model_inputs,
    max_new_tokens=256,
    do_sample=True,
    temperature=0.01,
    num_return_sequences=1
)
output_ids = generated_ids[0][len(model_inputs.input_ids[0]):].tolist()
output_text = tokenizer.decode(output_ids, skip_special_tokens=True)

print("Input:", input_txt)
print("Translation:", output_text)
````
